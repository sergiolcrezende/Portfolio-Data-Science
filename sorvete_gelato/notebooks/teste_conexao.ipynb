```python
from azure.identity import DefaultAzureCredential
from azure.mgmt.resource import ResourceManagementClient
import os

# Tenta obter as credenciais
credential = DefaultAzureCredential()

# Substitua pelo seu Subscription ID para testar
subscription_id = "7c42a7df-8885-4f11-beb2-87442a4a78ed"

try:
    resource_client = ResourceManagementClient(credential, subscription_id)
    # Tenta listar grupos de recursos para validar a conexão
    for group in resource_client.resource_groups.list():
        print(f"Conectado! Grupo encontrado: {group.name}")
        break
except Exception as e:
    print(f"Erro de conexão: {e}")
```
